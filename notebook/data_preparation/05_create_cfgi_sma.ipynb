```python
from pathlib import Path

import pandas as pd


# Paths
INPUT_PATH = Path("data/processed/base_dataframe_CFGI.csv")
OUTPUT_PATH = Path("data/processed/sma_cfgi_features.csv")


# Load the CFGI-augmented dataset
df = pd.read_csv(INPUT_PATH)


# Create 3-, 5-, and 7-day trailing moving averages of the CFGI
window_sizes = [3, 5, 7]

sma_df = pd.DataFrame()

for window in window_sizes:
    sma_df[f"value_sma{window}"] = (
        df["value"]
        .rolling(window=window)
        .mean()
    )


# Keep the initial NaN values generated by the rolling windows.
# They are handled later in the modeling pipeline.
print("First rows:")
print(sma_df.head(10))

print("\nMissing values:")
print(sma_df.isna().sum())


# Save CFGI moving-average features
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

sma_df.to_csv(
    OUTPUT_PATH,
    index=False
)

print(f"\nSaved CFGI moving-average features to: {OUTPUT_PATH}")
```
