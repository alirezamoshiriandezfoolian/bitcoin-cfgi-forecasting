```python
from pathlib import Path

import pandas as pd


# Paths
INPUT_PATH = Path("data/processed/base_dataframe_CFGI.csv")
OUTPUT_PATH = Path("data/processed/new_features_momentum.csv")


# Load the CFGI-augmented dataset
df = pd.read_csv(INPUT_PATH)


# Create one-day momentum features
df["price_momentum"] = (
    df["close"] - df["close"].shift(1)
)

df["volume_momentum"] = (
    df["volume"] - df["volume"].shift(1)
)

df["RS_volatility_momentum"] = (
    df["rs_volatility"] - df["rs_volatility"].shift(1)
)

df["CFGI_momentum"] = (
    df["value"] - df["value"].shift(1)
)


# Inspect the resulting momentum features
momentum_columns = [
    "price_momentum",
    "volume_momentum",
    "RS_volatility_momentum",
    "CFGI_momentum"
]

print("First rows:")
print(df[momentum_columns].head())

print("\nMissing values:")
print(df[momentum_columns].isna().sum())


# Save the dataset with momentum features
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

df.to_csv(
    OUTPUT_PATH,
    index=False
)

print(f"\nSaved momentum features to: {OUTPUT_PATH}")
```
