```python
from pathlib import Path

import pandas as pd


# Paths
INPUT_PATH = Path("data/processed/base_dataframe_CFGI.csv")
OUTPUT_PATH = Path("data/processed/lag_cfgi_lags.csv")


# Load the CFGI-augmented dataset
df = pd.read_csv(INPUT_PATH)


# Create backward-looking CFGI lags from 1 to 7 days
lagged_df = pd.DataFrame()

for lag in range(1, 8):
    lagged_df[f"value_lag{lag}"] = df["value"].shift(lag)


# Keep the initial NaN values generated by lagging.
# They are handled later in the modeling pipeline.
print("First rows:")
print(lagged_df.head(10))

print("\nMissing values:")
print(lagged_df.isna().sum())


# Save the CFGI lag features
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

lagged_df.to_csv(
    OUTPUT_PATH,
    index=False
)

print(f"\nSaved CFGI lag features to: {OUTPUT_PATH}")
```
