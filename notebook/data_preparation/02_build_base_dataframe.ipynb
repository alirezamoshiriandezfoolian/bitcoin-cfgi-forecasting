```python
from pathlib import Path

import pandas as pd


# Paths
INPUT_PATH = Path("data/processed/bitcoin_with_volatility.csv")
OUTPUT_PATH = Path("data/processed/base_dataframe.csv")


# Load Bitcoin data with Rogers-Satchell volatility
df = pd.read_csv(INPUT_PATH)


# Standardize column names
df.columns = df.columns.str.lower()


# Check available columns
print("Available columns:")
print(df.columns.tolist())


# Keep only the variables used in the baseline model
selected_columns = [
    "timeopen",
    "close",
    "volume",
    "rs_volatility"
]

df_selected = df[selected_columns].copy()


# Remove observations with zero Rogers-Satchell volatility
df_selected = df_selected[
    df_selected["rs_volatility"] != 0
].copy()


# Inspect the resulting dataset
print("\nFirst rows:")
print(df_selected.head())

print("\nShape:")
print(df_selected.shape)


# Save the baseline dataset
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
df_selected.to_csv(OUTPUT_PATH, index=False)

print(f"\nSaved baseline dataset to: {OUTPUT_PATH}")
```
