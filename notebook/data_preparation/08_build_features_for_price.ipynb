```python
from pathlib import Path

import pandas as pd


# Paths
INPUT_PATH = Path("data/processed/merged_all_features.csv")
OUTPUT_PATH = Path("data/processed/features_for_price.csv")


# Load the complete feature dataset
df = pd.read_csv(INPUT_PATH)


# Select the variables used in the final
# CFGI + derived-features price-forecasting specification
selected_columns = [
    "timeopen",
    "close",
    "rs_volatility",
    "volume",
    "value",
    "CFGI_momentum",
    "value_lag1",
    "value_lag2",
    "value_lag3",
    "value_lag4",
    "value_lag5",
    "value_lag6",
    "value_lag7",
    "value_sma3",
    "value_sma5",
    "value_sma7"
]

features_for_price = df[selected_columns].copy()


# Inspect the final modeling dataset
print("Final dataset shape:", features_for_price.shape)

print("\nColumns:")
print(features_for_price.columns.tolist())

print("\nMissing values:")
print(features_for_price.isna().sum())


# Save the final derived-feature dataset
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

features_for_price.to_csv(
    OUTPUT_PATH,
    index=False
)

print(f"\nSaved final price-feature dataset to: {OUTPUT_PATH}")
```
