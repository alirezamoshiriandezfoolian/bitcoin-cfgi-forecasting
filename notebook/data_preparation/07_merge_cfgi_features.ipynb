```python
from pathlib import Path

import pandas as pd


# Paths
BASE_INPUT_PATH = Path("data/processed/base_dataframe_CFGI.csv")
MOMENTUM_INPUT_PATH = Path("data/processed/new_features_momentum.csv")
LAG_INPUT_PATH = Path("data/processed/lag_cfgi_lags.csv")
SMA_INPUT_PATH = Path("data/processed/sma_cfgi_features.csv")

OUTPUT_PATH = Path("data/processed/merged_all_features.csv")


# Load the two base columns that should appear only once
df_main = pd.read_csv(
    BASE_INPUT_PATH
)[["timeopen", "rs_volatility"]]


# Load momentum features
df_momentum = pd.read_csv(MOMENTUM_INPUT_PATH)

# Remove columns already supplied by df_main
df_momentum = df_momentum.drop(
    columns=["timeopen", "rs_volatility"]
)


# Load CFGI lag and moving-average features
df_lags = pd.read_csv(LAG_INPUT_PATH)
df_sma = pd.read_csv(SMA_INPUT_PATH)


# Merge all feature groups by their original row order
df_merged = pd.concat(
    [
        df_main,
        df_momentum,
        df_lags,
        df_sma
    ],
    axis=1
)


# Inspect the resulting dataset
print("Merged dataset shape:", df_merged.shape)

print("\nColumns:")
print(df_merged.columns.tolist())

print("\nMissing values:")
print(df_merged.isna().sum())


# Save the complete feature dataset
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

df_merged.to_csv(
    OUTPUT_PATH,
    index=False
)

print(f"\nSaved merged feature dataset to: {OUTPUT_PATH}")
```
